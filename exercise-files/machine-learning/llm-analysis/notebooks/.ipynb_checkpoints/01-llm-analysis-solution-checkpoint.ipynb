```python
from openai import OpenAI
import pandas as pd
```

#### Adding your API key
Then you need to open up an OpenA client using your API key. An API key is akin to a password. You can create an API key on OpenAI's [website](https://platform.openai.com/api-keys). Using OpenAI's API costs money and the company usually requires you to purchase credits to be able to use it. 

Once you have an API key and added credits, you can paste it into the string below next to `api_key=`. Note that `rate limits`, i.e. the number of times you can run your code and access the API may vary. You can read more about rate limits [here](https://platform.openai.com/docs/guides/rate-limits).

```python
# Initialize the client with your API key
client = OpenAI(api_key="") # Replace with your actual API key
```

#### Calling the API 
The lines below help you call the API key. 

```python
# This creates a chat completion request
response = client.chat.completions.create(
     # this specifies which model you will use:
    model="gpt-3.5-turbo",
    messages=[
         # this gives instructions on how the model should behave:
        {"role": "system", "content": "You are a helpful assistant."},
         # this gives instructions on that the model should do:
        {"role": "user", "content": "Tell me a fun fact about space."}
    ]
)
```

Then you can print the model's response:

```python
print(response.choices[0].message.content)
```

#### Analysis of speeches 

The data used in this exercise is a pared down version from this [data set of 900+ speeches made by U.S. Presidents](https://docs.google.com/spreadsheets/d/1YUuWcioqwX7kOx6CI278J2n25D2nVdLOHvkxlWNd9Ik/edit?gid=1455551723#gid=1455551723). It was downloaded from [Kaggle](http://kaggle.com/datasets/kboghe/presidentialspeeches?resource=download) and represents Presidential speeches given between 1789 and 2020. The data is a randomized sample of 5 speeches. 


```python
speeches = pd.read_csv("../data/presidential_speeches - radomized sample.csv")
speeches.head()
```

Make a function that responds to the content of each speech:

```python
def getresponse(x):
    response = client.chat.completions.create(
         # this specifies which model you will use:
        model="gpt-3.5-turbo",
        messages=[
             # this gives instructions on how the model should behave:
            {"role": "system", "content": "You are helping me identify a presidential speech."},
             # this gives instructions on that the model should do:
            {"role": "user", "content": f"Here is a document: {x}\n\nTell me what year this speech is from and return only the full year."}
            ]
    )
    return response.choices[0].message.content
```

Create a new column and return the instructions set in the function in that column: 

```python
speeches["approximate_year"] = speeches["speech"].apply(getresponse)
```
